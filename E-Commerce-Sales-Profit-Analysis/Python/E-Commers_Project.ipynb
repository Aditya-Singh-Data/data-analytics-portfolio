```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    origin="1899-12-30",
    unit="D"
)
```

```python
df["Order Date"].head()
```

```python
df["Month"]=df["Order Date"].dt.month
```

```python
# Q1 Calculate the monthly sales of the store and identify which month had
# the highest sales and which month had the lowest sales

Monthly_wise_sales = df.groupby("Month", as_index=True)["Sales"].sum()

# Highest and lowest sales month
highest_month = Monthly_wise_sales.idxmax()
lowest_month = Monthly_wise_sales.idxmin()

# Highest and lowest sales value
highest_sales = Monthly_wise_sales.max()
lowest_sales = Monthly_wise_sales.min()


print("Highest Sales Month:", highest_month)
print("Highest Sales:", highest_sales)

print("Lowest Sales Month:", lowest_month)
print("Lowest Sales:", lowest_sales)




month_names = {
    1: "Jan",
    2: "Feb",
    3: "Mar",
    4: "Apr",
    5: "May",
    6: "Jun",
    7: "Jul",
    8: "Aug",
    9: "Sep",
    10: "Oct",
    11: "Nov",
    12: "Dec"
}

Monthly_wise_sales.index = Monthly_wise_sales.index.map(month_names)

# Line Chart
plt.figure(figsize=(9,3))

plt.plot(
    Monthly_wise_sales.index,
    Monthly_wise_sales.values,
    marker="o"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(range(1, 13))
plt.grid(True)

plt.show()
```

```python
# Q2 Analyze sales based on product categories and determine which category has the lowest sales and which category has the highest sales

categories_wise_Sales= df.groupby("Category")["Sales"].sum()

Highest_sales_categories = categories_wise_Sales.idxmax()
Lowest_sales_categorie = categories_wise_Sales.idxmin()


Highest_sales = categories_wise_Sales.max()
Lowest_sales  = categories_wise_Sales.min()

print("Highest Sales Categories :", Highest_sales_categories)
print("Highest Sales:",Highest_sales)

print("Lowest Sales Categorie :",Lowest_sales_categorie  )
print("Lowest Sales:", Lowest_sales)






plt.figure(figsize=(9,3))

plt.bar(categories_wise_Sales.index,
        categories_wise_Sales.values
        
        )


plt.title("categories_wise_Sales")
plt.xlabel('Categorie')
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python
# Q3 Sales analysis based on sub-categories



sub_categories_wise_Sales= df.groupby("Sub-Category")["Sales"].sum()



Highest_sales_sub_categories = sub_categories_wise_Sales.idxmax()
Lowest_sales_sub_categorie = sub_categories_wise_Sales.idxmin()


Highest_sales = sub_categories_wise_Sales.max()
Lowest_sales  = sub_categories_wise_Sales.min()



print("Highest Sales Sub-Categories :", Highest_sales_sub_categories)
print("Highest Sales:",Highest_sales)

print("Lowest Sales Sub-Categorie :",Lowest_sales_sub_categorie  )
print("Lowest Sales:", Lowest_sales)


  

plt.figure(figsize=(20, 7))

plt.bar(
    sub_categories_wise_Sales.index,
    sub_categories_wise_Sales.values
)

plt.title("Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")
plt.grid(axis="y")

plt.show()
```

```python
# Q4 Analyze monthly profit from sales and determine which month had the highest profit

Monthly_wise_profit = df.groupby("Month")["Profit"].sum().round(2)

highest_profit_Monthly = Monthly_wise_profit.idxmax()

highest_profit = Monthly_wise_profit.max()


print("Highest Profit Monthly:",highest_profit_Monthly)
print("Highes Profit :" , highest_profit)

month_names = {
    1: "Jan",
    2: "Feb",
    3: "Mar",
    4: "Apr",
    5: "May",
    6: "Jun",
    7: "Jul",
    8: "Aug",
    9: "Sep",
    10: "Oct",
    11: "Nov",
    12: "Dec"
}

Monthly_wise_profit.index = Monthly_wise_profit.index.map(month_names)



plt.plot(
    Monthly_wise_profit.index,
    Monthly_wise_profit.values,
    marker="o"
)

plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")

plt.xticks(
    range(1, 13),
    [month_names[i] for i in range(1, 13)]
)

plt.grid()
plt.show()
```

```python
# Q5 Analyze profit by category and sub-category

category_and_sub_category_wise_profit = (
    df.groupby(["Category", "Sub-Category"])["Profit"]
    .sum()
    .reset_index()
)

print(category_and_sub_category_wise_profit)


category_and_sub_category_wise_profit["Category_Sub_Category"] = (
    category_and_sub_category_wise_profit["Category"]
    + " - "
    + category_and_sub_category_wise_profit["Sub-Category"]
)

plt.figure(figsize=(10, 8))

plt.barh(
    category_and_sub_category_wise_profit["Category_Sub_Category"],
    category_and_sub_category_wise_profit["Profit"]
)

plt.title("Profit by Category and Sub-Category")
plt.xlabel("Profit")
plt.ylabel("Category - Sub-Category")

plt.tight_layout()
plt.show()
```

```python
# Q6 Analyze sales and profit by customer segment


customer_segment_wise_Sales_and_Profit = df.groupby("Segment")[["Sales" , "Profit"]].sum()

print(customer_segment_wise_Sales_and_Profit)


plt.figure(figsize=(8,5))

customer_segment_wise_Sales_and_Profit.plot(
    kind="bar"
)

plt.title("Sales and Profit by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Amount")

plt.tight_layout()
plt.show()

```

```python

# Q7 Analyze sales to profit ratio


total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

sales_profit_ratio = total_sales / total_profit

print("Sales to Profit Ratio:", sales_profit_ratio.round(2))

sales_profit = {
    "Sales": total_sales,
    "Profit": total_profit
}

plt.bar(
    sales_profit.keys(),
    sales_profit.values()
)

plt.title("Total Sales vs Total Profit")
plt.xlabel("Metric")
plt.ylabel("Amount")
plt.show()
```
